```python
# What types of health facilities do we have?
if 'amenity' in facilities.columns:
    print(facilities['amenity'].value_counts())
elif 'healthcare' in facilities.columns:
    print(facilities['healthcare'].value_counts())
else:
    # Print all columns with their unique values to find the type column
    for col in facilities.columns:
        if facilities[col].dtype == 'object':
            print(f"\n{col}:\n{facilities[col].value_counts().head(10)}")
```

```python
cols_to_drop = [
    'addr_full', 'addr_city', 'capacity_persons', 'source',
    'adm4_pcode', 'adm4_name', 'building', 'healthcare_speciality',
    'name_rw', 'name_fr', 'name_latin', 'name_en'
]
facilities_clean = facilities.drop(columns=cols_to_drop)
facilities_clean = facilities_clean.dropna(subset=['amenity'])
print(f"Rows after dropping missing amenity: {len(facilities_clean)}")
```

```python
facilities_clean = facilities_clean.to_crs(epsg=4326)
districts = districts.to_crs(epsg=4326)
provinces = provinces.to_crs(epsg=4326)
sectors   = sectors.to_crs(epsg=4326)
print(f"Facilities CRS: {facilities_clean.crs}")
```

```python
primary = ['hospital', 'clinic', 'health_post', 'doctors']
support = ['pharmacy', 'drugstore', 'dentist']

def categories(amenity):
    if amenity in primary:
        return 'primary'
    elif amenity in support:
        return 'support'
    else:
        return 'other'
facilities_clean['category']=facilities["amenity"].apply(categories)
```

```python
missing_adm = facilities_clean['adm2_name'].isnull()
if missing_adm.sum() > 0:
    joined = gpd.sjoin(
        facilities_clean[missing_adm][['geometry']],
        districts[['ADM2_EN', 'ADM1_EN', 'geometry']],  
        how='left', predicate='within'
    )
    # We'll handle column name mismatch after seeing district columns
    print(f"Facilities missing district info: {missing_adm.sum()}")
```

```python
districts.columns
```

```python
print(f"\nClean facilities shape: {facilities_clean.shape}")
print(f"\nAmenity breakdown:\n{facilities_clean['amenity'].value_counts()}")
print(f"\nCategory breakdown:\n{facilities_clean['category'].value_counts()}")
print(f"\nRemaining columns: {facilities_clean.columns.tolist()}")
```

```python
# ── Check district column names first ─────────────────────────────────────
print("District columns:", districts.columns.tolist())
```

```python
facilities_joined = gpd.sjoin(
    facilities_clean,
    districts[['ADM1_EN', 'ADM2_EN', 'ADM2_PCODE', 'geometry']],
    how='left',
    predicate='within'
)

# Drop the index column added by sjoin
facilities_joined = facilities_joined.drop(columns=['index_right'])
```

```python
# ── Check how many facilities didn't fall within any district ──────────────
unmatched = facilities_joined['ADM2_EN'].isnull().sum()
print(f"Facilities matched to a district : {facilities_joined['ADM2_EN'].notna().sum()}")
print(f"Facilities NOT matched (outside boundaries): {unmatched}")
```

```python
if unmatched > 0:
    unmatched_fac = facilities_clean[facilities_joined['ADM2_EN'].isnull()].copy()
    nearest = gpd.sjoin_nearest(
        unmatched_fac,
        districts[['ADM1_EN', 'ADM2_EN', 'ADM2_PCODE', 'geometry']],
        how='left'
    ).drop(columns=['index_right'])
    
    facilities_joined.loc[facilities_joined['ADM2_EN'].isnull(), 
                          ['ADM1_EN', 'ADM2_EN', 'ADM2_PCODE']] = \
        nearest[['ADM1_EN', 'ADM2_EN', 'ADM2_PCODE']].values
    
    print(f"Fixed {unmatched} unmatched facilities using nearest district")
```

```python
print(f"\nFacilities per district (sample):")
print(facilities_joined['ADM2_EN'].value_counts().head(10))
```

```python
facilities_joined.head()
```

```python
facility_counts = (
    facilities_joined
    .groupby('ADM2_EN')
    .agg(
        total_facilities   = ('amenity', 'count'),
        hospitals          = ('amenity', lambda x: (x == 'hospital').sum()),
        clinics            = ('amenity', lambda x: (x == 'clinic').sum()),
        pharmacies         = ('amenity', lambda x: (x == 'pharmacy').sum()),
        primary_care       = ('category', lambda x: (x == 'primary').sum()),
        support            = ('category', lambda x: (x == 'support').sum()),
    )
    .reset_index()
    .sort_values('total_facilities', ascending=False)
)
facility_counts.head(10)
```

```python
districts_analysis = districts.merge(facility_counts, on='ADM2_EN', how='left')
districts_analysis['total_facilities'] = districts_analysis['total_facilities'].fillna(0)
districts_analysis.head(10)
```

```python
sorted_districts = districts_analysis[['ADM2_EN','total_facilities', 
                                        'hospitals', 'clinics', 'pharmacies', 'primary_care']] \
                    .sort_values('total_facilities', ascending=False)

print("Top 10 districts by total facilities:")
print(sorted_districts.head(10).to_string(index=False))

print("\nBottom 10 districts by total facilities:")
print(sorted_districts.tail(10).to_string(index=False))

print(f"\nTotal facilities across all districts: {sorted_districts['total_facilities'].sum()}")
print(f"Average per district: {sorted_districts['total_facilities'].mean():.1f}")
print(f"Max: {sorted_districts['total_facilities'].max()} | Min: {sorted_districts['total_facilities'].min()}")
```

```python
from rasterstats import zonal_stats

# ── Make sure districts are in same CRS as raster ─────────────────────────
print(f"Raster CRS : {pop_raster.crs}")
print(f"Districts CRS: {districts_analysis.crs}")

# Reproject districts to match raster CRS if needed
districts_proj = districts_analysis.to_crs(pop_raster.crs)
```

```python
print("\nCalculating population per district")

stats = zonal_stats(
    districts_proj,
    "rwa_pop_2025_CN_100m_R2025A_v1.tif",
    stats=['sum', 'mean', 'max'],
    nodata=-99999
)
stats
```

```python
#Add results to dataframe
districts_analysis['population']     = [s['sum']  for s in stats]
districts_analysis['pop_density_avg']= [s['mean'] for s in stats]
districts_analysis['pop_max_cell']   = [s['max']  for s in stats]
```

```python
# Calculate facilities per 10,000 people 
districts_analysis['facility_per_10k'] = (
    districts_analysis['total_facilities'] / districts_analysis['population'] * 10000
)
districts_analysis['primary_per_10k'] = (
    districts_analysis['primary_care'] / districts_analysis['population'] * 10000
)

#Summary
result = districts_analysis[['ADM2_EN', 'population', 'total_facilities', 'facility_per_10k']] \
         .sort_values('facility_per_10k', ascending=False)

print("\nFacilities per 10,000 people (Top 10):")
print(result.head(10).to_string(index=False))
print("\nFacilities per 10,000 people (Bottom 10):")
print(result.tail(10).to_string(index=False))
print(f"\nNational average: {districts_analysis['facility_per_10k'].mean():.2f} per 10,000 people")
```

```python
from matplotlib.colors import Normalize
import matplotlib.cm as cm
import matplotlib.patches as mpatches

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# ── Left: Facilities per 10k ───────────────────────────────────────────────
districts_analysis.plot(
    column='facility_per_10k',
    ax=axes[0],
    cmap='RdYlGn',
    legend=True,
    legend_kwds={'label': 'Facilities per 10,000 people', 'shrink': 0.6},
    edgecolor='black',
    linewidth=0.5
)

for _, row in districts_analysis.iterrows():
    axes[0].annotate(
        f"{row['ADM2_EN']}\n{row['facility_per_10k']:.2f}",
        xy=(row.geometry.centroid.x, row.geometry.centroid.y),
        fontsize=5, ha='center', color='white'
    )

axes[0].set_title('Facilities per 10,000 People\n(Coverage Rate)', 
                   fontsize=13, fontweight='bold', pad=10)
axes[0].axis('off')

# ── Right: Raw count vs per capita side-by-side comparison ────────────────
# Normalize both columns for comparison


# Classify districts by coverage level
def coverage_class(val):
    if val >= 2.0:
        return 'High (≥2.0)'
    elif val >= 1.0:
        return 'Medium (1.0–2.0)'
    elif val >= 0.5:
        return 'Low (0.5–1.0)'
    else:
        return 'Critical (<0.5)'

districts_analysis['coverage_class'] = districts_analysis['facility_per_10k'].apply(coverage_class)

color_map = {
    'High (≥2.0)'     : '#1a9641',
    'Medium (1.0–2.0)': '#a6d96a',
    'Low (0.5–1.0)'   : '#fdae61',
    'Critical (<0.5)' : '#d7191c'
}

districts_analysis['color'] = districts_analysis['coverage_class'].map(color_map)
districts_analysis.plot(
    ax=axes[1],
    color=districts_analysis['color'],
    edgecolor='black',
    linewidth=0.5
)

for _, row in districts_analysis.iterrows():
    axes[1].annotate(
        row['ADM2_EN'],
        xy=(row.geometry.centroid.x, row.geometry.centroid.y),
        fontsize=5, ha='center', color='black'
    )

# Manual legend

legend_patches = [mpatches.Patch(color=c, label=l) for l, c in color_map.items()]
axes[1].legend(handles=legend_patches, loc='lower left', fontsize=8, title='Coverage Level')
axes[1].set_title('Coverage Classification by District', 
                   fontsize=13, fontweight='bold', pad=10)
axes[1].axis('off')

plt.suptitle('Rwanda — Health Facility Coverage per Capita', 
             fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

```python
import numpy as np

# ── Project to UTM zone 36S (meters) for accurate distances ───────────────
districts_utm  = districts_analysis.to_crs(epsg=32736)
facilities_utm = facilities_joined.to_crs(epsg=32736)

# ── Get district centroids as a GeoDataFrame ───────────────────────────────
centroids_gdf = districts_utm.copy()
centroids_gdf['geometry'] = districts_utm.geometry.centroid

# ── Find nearest facility to each centroid using sjoin_nearest ─────────────
nearest = gpd.sjoin_nearest(
    centroids_gdf[['ADM2_EN', 'geometry']],
    facilities_utm[['name', 'amenity', 'geometry']],
    how='left',
    distance_col='nearest_fac_m'
)

# Keep only first match per district (in case of ties)
nearest = nearest.drop_duplicates(subset='ADM2_EN')

# ── Add results back to districts_analysis ────────────────────────────────
districts_analysis['nearest_fac_km']   = nearest['nearest_fac_m'].values / 1000
districts_analysis['nearest_fac_name'] = nearest['name'].values
districts_analysis['nearest_fac_type'] = nearest['amenity'].values

print("✅ Done!")
print(districts_analysis[['ADM2_EN', 'nearest_fac_km', 'nearest_fac_type']]
      .sort_values('nearest_fac_km', ascending=False)
      .to_string(index=False))
```

```python
# ── Normalize each metric to 0-1 scale ────────────────────────────────────
# For coverage rate: lower = worse, so we invert it
# For distance: higher = worse
# For population: higher = more people affected = worse

def normalize(series, invert=False):
    min_val = series.min()
    max_val = series.max()
    normalized = (series - min_val) / (max_val - min_val)
    if invert:
        normalized = 1 - normalized
    return normalized

districts_analysis['score_coverage'] = normalize(districts_analysis['facility_per_10k'], invert=True)
districts_analysis['score_distance']  = normalize(districts_analysis['nearest_fac_km'], invert=False)
districts_analysis['score_population']= normalize(districts_analysis['population'], invert=False)

# ── Weighted combined score ────────────────────────────────────────────────
# Coverage rate = 40%, Distance = 40%, Population = 20%
districts_analysis['vulnerability_score'] = (
    districts_analysis['score_coverage']  * 0.40 +
    districts_analysis['score_distance']  * 0.40 +
    districts_analysis['score_population']* 0.20
)

# ── Rank districts ─────────────────────────────────────────────────────────
districts_analysis['vulnerability_rank'] = (
    districts_analysis['vulnerability_score']
    .rank(ascending=False)
    .astype(int)
)

# ── Print final ranking ────────────────────────────────────────────────────
final_ranking = districts_analysis[[
    'ADM2_EN', 'vulnerability_rank', 'vulnerability_score',
    'facility_per_10k', 'nearest_fac_km', 'population', 'coverage_class'
]].sort_values('vulnerability_rank')

print("=" * 75)
print("RWANDA HEALTH FACILITY VULNERABILITY RANKING (1 = most at risk)")
print("=" * 75)
final_ranking
```

```python
import networkx as nx
# ── Convert to NetworkX graph ──────────────────────────────────────────────
print("Building graph...")
G = osm.to_graph(nodes, edges, graph_type="networkx")
print(f"✅ Graph built")
print(f"   Nodes : {G.number_of_nodes()}")
print(f"   Edges : {G.number_of_edges()}")
```

```python
from shapely.geometry import Point

# ── Reproject everything to WGS84 ─────────────────────────────────────────
nodes_gdf = nodes.to_crs(epsg=4326)

# Get node coordinates as array
node_coords = np.array([(row.geometry.x, row.geometry.y) 
                         for _, row in nodes_gdf.iterrows()])
node_ids    = nodes_gdf['id'].values

# ── Function to find nearest node to a point ──────────────────────────────
from scipy.spatial import cKDTree

node_tree = cKDTree(node_coords)

def nearest_node(lon, lat):
    dist, idx = node_tree.query([lon, lat], k=1)
    return node_ids[idx]

# ── Snap each facility to nearest road node ────────────────────────────────
print("Snapping facilities to road network...")
facilities_utm4326 = facilities_joined.to_crs(epsg=4326)

facilities_joined['nearest_node'] = [
    nearest_node(row.geometry.x, row.geometry.y)
    for _, row in facilities_utm4326.iterrows()
]

# ── Snap each district centroid to nearest road node ──────────────────────
print("Snapping district centroids to road network...")
districts_4326 = districts_analysis.to_crs(epsg=4326)

districts_analysis['centroid_node'] = [
    nearest_node(row.geometry.centroid.x, row.geometry.centroid.y)
    for _, row in districts_4326.iterrows()
]

print("✅ Done snapping!")
print(f"Sample facility nodes: {facilities_joined['nearest_node'].head().values}")
```

```python
# ── Add travel time weights to edges ──────────────────────────────────────
WALKING_SPEED_MPM = 83.3

for u, v, data in G.edges(data=True):
    length = data.get('length', 0) or 0
    data['travel_time_min'] = length / WALKING_SPEED_MPM

print("Travel time weights added")

# ── Get all unique facility nodes that EXIST in the graph ─────────────────
all_facility_nodes = list(set(facilities_joined['nearest_node'].values))
facility_nodes     = [n for n in all_facility_nodes if n in G]

print(f"Total facility nodes    : {len(all_facility_nodes)}")
print(f"Valid nodes in graph    : {len(facility_nodes)}")
print(f"Filtered out            : {len(all_facility_nodes) - len(facility_nodes)}")

# ── Multi-source Dijkstra ──────────────────────────────────────────────────
print("\nRunning shortest path analysis")

distances_to_facility = nx.multi_source_dijkstra_path_length(
    G,
    sources=facility_nodes,
    weight='travel_time_min'
)

print("Shortest paths calculated!")
```

```python
district_travel_times = []

for _, row in districts_analysis.iterrows():
    centroid_node = row['centroid_node']
    district_name = row['ADM2_EN']

    # Check if centroid node is in graph AND reachable
    if centroid_node in G and centroid_node in distances_to_facility:
        travel_time = distances_to_facility[centroid_node]
    else:
        # Fallback: straight-line distance converted to walking time
        travel_time = (row['nearest_fac_km'] * 1000) / WALKING_SPEED_MPM
        print(f"  ⚠️  {district_name}: using straight-line fallback")

    district_travel_times.append({
        'ADM2_EN'         : district_name,
        'travel_time_min' : round(travel_time, 1),
        'straight_line_km': round(row['nearest_fac_km'], 2)
    })

travel_df = pd.DataFrame(district_travel_times).sort_values('travel_time_min', ascending=False)

print("=" * 60)
print("ACTUAL WALKING TIME TO NEAREST HEALTH FACILITY")
print("=" * 60)
print(travel_df.to_string(index=False))
print(f"\nNational average walking time : {travel_df['travel_time_min'].mean():.1f} minutes")
print(f"Worst district : {travel_df.iloc[0]['ADM2_EN']} — {travel_df.iloc[0]['travel_time_min']:.1f} min")
print(f"Best district  : {travel_df.iloc[-1]['ADM2_EN']} — {travel_df.iloc[-1]['travel_time_min']:.1f} min")
```

```python
# ── Merge travel times into districts_analysis ────────────────────────────
districts_analysis = districts_analysis.merge(
    travel_df[['ADM2_EN', 'travel_time_min']], 
    on='ADM2_EN', 
    how='left'
)

print("Travel times added to districts_analysis")
print(districts_analysis[['ADM2_EN', 'nearest_fac_km', 'travel_time_min']].head(10))
```

```python
import plotly.express as px
import plotly.graph_objects as go
import json

DARK_BG    = '#0f1117'
CARD_BG    = '#1a1d27'
PANEL_BG   = '#1e2130'
ACCENT     = '#4e9af1'
TEXT_COLOR = '#e0e0e0'
GRID_COLOR = '#2a2d3e'

# ── No legend here — set per chart ────────────────────────────────────────
plotly_dark = dict(
    plot_bgcolor  = CARD_BG,
    paper_bgcolor = CARD_BG,
    font          = dict(color=TEXT_COLOR, family='Segoe UI'),
    xaxis         = dict(gridcolor=GRID_COLOR, linecolor=GRID_COLOR, tickcolor=GRID_COLOR),
    yaxis         = dict(gridcolor=GRID_COLOR, linecolor=GRID_COLOR, tickcolor=GRID_COLOR),
    title         = dict(font=dict(color=TEXT_COLOR, size=14), x=0.02)
)

LEGEND_BOTTOM = dict(
    bgcolor='rgba(0,0,0,0)',
    font=dict(color=TEXT_COLOR),
    orientation='h',
    yanchor='bottom',
    y=-0.18,
    x=0
)

cov_colors = {
    'High (≥2.0)'     : '#2ecc71',
    'Medium (1.0–2.0)': '#a8e063',
    'Low (0.5–1.0)'   : '#f39c12',
    'Critical (<0.5)' : '#e74c3c'
}

# ── Figure 1: Coverage choropleth ─────────────────────────────────────────
fig_map = px.choropleth_mapbox(
    dashboard_df,
    geojson=geojson_data,
    locations=dashboard_df.index,
    color='facility_per_10k',
    hover_name='ADM2_EN',
    hover_data={
        'facility_per_10k'          : ':.2f',
        'total_facilities'     : True,
        'population'           : ':,',
        'travel_time_min'      : ':.1f',
        'coverage_class'       : True,
        'vulnerability_rank_v2': True
    },
    color_continuous_scale=[
        [0.0, '#e74c3c'], [0.3, '#f39c12'],
        [0.6, '#a8e063'], [1.0, '#2ecc71']
    ],
    mapbox_style='carto-darkmatter',
    zoom=7,
    center={'lat': -1.9403, 'lon': 29.8739},
    opacity=0.8,
    labels={
        'facility_per_10k'          : 'Per 10k',
        'total_facilities'     : 'Total',
        'population'           : 'Population',
        'travel_time_min'      : 'Walk (min)',
        'coverage_class'       : 'Coverage',
        'vulnerability_rank_v2': 'Vuln. Rank'
    }
)
fig_map.update_layout(
    height=480,
    margin=dict(l=0, r=0, t=10, b=0),
    paper_bgcolor=CARD_BG,
    coloraxis_colorbar=dict(
        title=dict(text='Per 10k', font=dict(color=TEXT_COLOR)),
        tickfont=dict(color=TEXT_COLOR)
    )
)

# ── Figure 2: Vulnerability ranking ───────────────────────────────────────
vuln_sorted = dashboard_df.sort_values('vulnerability_score_v2', ascending=True)

fig_vuln = px.bar(
    vuln_sorted,
    x='vulnerability_score_v2',
    y='ADM2_EN',
    color='coverage_class',
    color_discrete_map=cov_colors,
    orientation='h',
    hover_data={
        'travel_time_min'      : ':.1f',
        'facility_per_10k'          : ':.2f',
        'population'           : ':,',
        'vulnerability_rank_v2': True
    },
    labels={
        'vulnerability_score_v2': 'Vulnerability Score',
        'ADM2_EN'               : '',
        'coverage_class'        : 'Coverage',
        'travel_time_min'       : 'Walk (min)',
        'facility_per_10k'           : 'Per 10k',
        'population'            : 'Population',
        'vulnerability_rank_v2' : 'Rank'
    }
)
fig_vuln.update_layout(
    **plotly_dark,
    height=580,
    xaxis_title='Vulnerability Score',
    yaxis_title='',
    margin=dict(l=10, r=20, t=20, b=40),
    legend=dict(
        orientation='h', yanchor='bottom',
        y=-0.12, x=0, font=dict(color=TEXT_COLOR)
    )
)
fig_vuln.update_traces(marker_line_width=0)

# ── Figure 3: Scatter — travel time vs coverage ───────────────────────────
fig_scatter = px.scatter(
    dashboard_df,
    x='travel_time_min',
    y='facility_per_10k',
    color='coverage_class',
    size='population',
    hover_name='ADM2_EN',
    color_discrete_map=cov_colors,
    size_max=45,
    labels={
        'travel_time_min': 'Walking Time (min)',
        'facility_per_10k'    : 'Facilities per 10,000',
        'coverage_class' : 'Coverage',
        'population'     : 'Population'
    }
)
fig_scatter.add_hline(
    y=1.0, line_dash='dash',
    line_color='rgba(255,255,255,0.3)',
    annotation_text='Acceptable threshold',
    annotation_font_color=TEXT_COLOR
)
fig_scatter.add_vline(
    x=45.3, line_dash='dash',
    line_color='rgba(255,165,0,0.4)',
    annotation_text='National avg',
    annotation_font_color='#f39c12'
)
fig_scatter.update_layout(
    **plotly_dark,
    height=420,
    margin=dict(l=10, r=20, t=20, b=40),
    legend=dict(
        orientation='h', yanchor='bottom',
        y=-0.18, x=0, font=dict(color=TEXT_COLOR)
    )
)
fig_scatter.update_traces(marker=dict(line=dict(width=0.5, color='rgba(0,0,0,0.3)')))

# ── Figure 4: Facility types stacked bar ──────────────────────────────────
top15 = districts_analysis.nlargest(15, 'total_facilities')

fig_types = go.Figure()
fig_types.add_trace(go.Bar(
    name='Hospitals',  x=top15['ADM2_EN'], y=top15['hospitals'],
    marker_color='#e74c3c', marker_line_width=0
))
fig_types.add_trace(go.Bar(
    name='Clinics',    x=top15['ADM2_EN'], y=top15['clinics'],
    marker_color='#4e9af1', marker_line_width=0
))
fig_types.add_trace(go.Bar(
    name='Pharmacies', x=top15['ADM2_EN'], y=top15['pharmacies'],
    marker_color='#2ecc71', marker_line_width=0
))
fig_types.update_layout(
    **plotly_dark,
    barmode='stack',
    height=380,
    xaxis_title='',
    yaxis_title='Number of Facilities',
    margin=dict(l=10, r=20, t=20, b=60),
    legend=dict(
        orientation='h', yanchor='bottom',
        y=1.02, x=0, font=dict(color=TEXT_COLOR)
    )
)

# ── Figure 5: Travel time bar chart ───────────────────────────────────────
travel_sorted = districts_analysis.sort_values('travel_time_min', ascending=False)

fig_travel = px.bar(
    travel_sorted,
    x='ADM2_EN',
    y='travel_time_min',
    color='coverage_class',
    color_discrete_map=cov_colors,
    labels={
        'travel_time_min': 'Walking Time (min)',
        'ADM2_EN'        : '',
        'coverage_class' : 'Coverage'
    }
)
fig_travel.add_hline(
    y=45.3, line_dash='dash',
    line_color='rgba(255,255,255,0.4)',
    annotation_text='National avg (45.3 min)',
    annotation_font_color=TEXT_COLOR
)
fig_travel.update_layout(
    **plotly_dark,
    height=380,
    xaxis_tickangle=-45,
    margin=dict(l=10, r=20, t=20, b=100),
    showlegend=False
)
fig_travel.update_traces(marker_line_width=0)

print("✅ All dark-themed figures ready")
```
